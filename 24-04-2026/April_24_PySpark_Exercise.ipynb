Section 8 — Window Functions

Exercises 71–80

```python
products.createOrReplaceTempView("products")
customers.createOrReplaceTempView("customers")
```

71. Rank products by price.

```python
spark.sql("""
SELECT *,
RANK() OVER(ORDER BY price DESC) AS price_rank
FROM products
""").show()
```

72. Rank products within category.

```python
spark.sql("""
SELECT *,
RANK() OVER(PARTITION BY category ORDER BY price DESC) AS category_rank
FROM products
""").show()
```

73. Row number for customers by signup_date.

```python
spark.sql("""
SELECT *,
ROW_NUMBER() OVER(ORDER BY signup_date) AS row_num
FROM customers
""").show()
```

74. Rank customers by revenue.

```python
spark.sql("""
SELECT c.name,
SUM(oi.quantity * p.price) AS total_revenue,
RANK() OVER(ORDER BY SUM(oi.quantity * p.price) DESC) AS rank
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
JOIN order_items oi ON o.order_id = oi.order_id
JOIN products p ON oi.product_id = p.product_id
GROUP BY c.name
""").show()
```

```python
final_df = spark.sql("""
SELECT
    c.customer_id,
    c.name,
    c.city,
    o.order_id,
    o.order_date,
    p.product_id,
    p.product_name,
    p.category,
    oi.quantity,
    p.price,
    oi.quantity * p.price AS revenue
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
JOIN order_items oi ON o.order_id = oi.order_id
JOIN products p ON oi.product_id = p.product_id
""")

final_df.createOrReplaceTempView("final_table")
```

75. Dense rank customers by revenue.

```python
spark.sql("""
SELECT
    name,
    SUM(revenue) AS total_revenue,
    DENSE_RANK() OVER (ORDER BY SUM(revenue) DESC) AS dense_rank
FROM final_table
GROUP BY name
""").show()
```

76. Find top customer per city.

```python
spark.sql("""
SELECT *
FROM (
    SELECT
        city,
        name,
        SUM(revenue) AS total_revenue,
        RANK() OVER (PARTITION BY city ORDER BY SUM(revenue) DESC) AS rank
    FROM final_table
    GROUP BY city, name
)
WHERE rank = 1
""").show()
```

77. Find top product per category.

```python
spark.sql("""
SELECT *
FROM (
    SELECT
        category,
        product_name,
        SUM(revenue) AS total_revenue,
        RANK() OVER (PARTITION BY category ORDER BY SUM(revenue) DESC) AS rank
    FROM final_table
    GROUP BY category, product_name
)
WHERE rank = 1
""").show()
```

78. Running total revenue by order date.

```python
spark.sql("""
SELECT
    order_date,
    SUM(revenue) AS daily_revenue,
    SUM(SUM(revenue)) OVER (ORDER BY order_date) AS running_total
FROM final_table
GROUP BY order_date
ORDER BY order_date
""").show()
```

79. Running total revenue by product.

```python
spark.sql("""
SELECT
    product_name,
    order_date,
    revenue,
    SUM(revenue) OVER (
        PARTITION BY product_name
        ORDER BY order_date
    ) AS running_total
FROM final_table
ORDER BY product_name, order_date
""").show()
```

80. Rank orders by payment amount.
