````python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("Employee Data").getOrCreate()
print("Spark Session Created")
````


### Exercise 1

Create a DataFrame with:

```
Name
Salary
Department
```

Add 5 employees.

````python
data = [
    ("Rahul", 60000, "IT"),
    ("Sneha", 45000, "HR"),
    ("Arjun", 70000, "IT"),
    ("Priya", 52000, "Finance"),
    ("Karan", 48000, "HR")
]
columns = ["Name", "Salary", "Department"]
df = spark.createDataFrame(data, columns)
df.show()
````

### Exercise 2

Show only **Name and Salary** columns.

````python
print("Name and Salary")
df.select("Name","Salary").show()
````

### Exercise 3

Filter employees with **salary greater than 50000**.


````python
print("Employees with salary greater than 50000")
df.filter(df.Salary>50000).show()
````

### Exercise 4

Count employees per department.


````python
print("Employee count per Department")
df.groupBy("Department").count().show()
````

### Exercise 5

Sort employees by salary.


````python
print("Sorted by salary (ascending)")
df.orderBy("Salary").show()
````

````python
print("Sorted by salary (descending)")
df.orderBy("Salary",ascending=False).show()
````
