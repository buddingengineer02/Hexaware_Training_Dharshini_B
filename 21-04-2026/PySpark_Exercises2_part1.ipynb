PART 1 — DataFrame Creation

Load all datasets into PySpark DataFrames.

Exercise 1 - Create a Spark session.

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("PySpark Exercise2").getOrCreate()
print("Spark Session Created")
```

Exercise 2 - Load employees.csv

```python
employees_df=spark.read.csv("employees.csv",header=True,inferSchema=True)
```

Exercise 3 - Load departments.csv

```python
departments_df = spark.read.csv("departments.csv", header=True, inferSchema=True)
```

Exercise 4 - Load sales.csv

```python
sales_df = spark.read.csv("sales.csv", header=True, inferSchema=True)
```

Exercise 5 - Load employees_nested.json

```python
nested_df = spark.read.json("employees_nested.json", multiLine=True)
```

Exercise 6 - Print schema for all DataFrames

```python
print("Employees Schema:")
employees_df.printSchema()

print("Departments Schema:")
departments_df.printSchema()

print("Sales Schema:")
sales_df.printSchema()

print("Nested JSON Schema:")
nested_df.printSchema()
```

PART 3 — Selecting and Renaming

Exercise 11 - Select only name and salary

```python
employees_df.select("name","salary").show()
```

Exercise 12 - Rename column salary to emp_salary

```python
employees_df.withColumnRenamed("salary","emp_salary").show()
```

Exercise 13 - Select employees who joined after 2023-01-01.

```python
employees_df.filter(employees_df.joining_date>"2023-01-01").show()
```

PART 4 — Filtering Data

Exercise 14 - Find employees with salary greater than 65000.

```python
employees_df.filter(employees_df.salary>65000).show()
```

Exercise 15 - Find employees in IT department.

```python
employees_df.filter(employees_df.department=="IT").show()
```

Exercise 16 - Find employees in IT department with salary greater than 70000.

```python
employees_df.filter((employees_df.department=="IT")&(employees_df.salary>70000)).show()
```

PART 5 — Dropping and Sorting

Exercise 17 - Drop column joining_date .

```python
employees_df.drop("joining_date").show()
```

Exercise 18 - Sort employees by salary ascending.

```python
employees_df.orderBy(employees_df.salary.asc()).show()
```

Exercise 19 - Sort employees by salary descending.

```python
employees_df.orderBy(employees_df.salary.desc()).show()
```

Exercise 20 - Find top 3 highest paid employees.

```python
employees_df.orderBy(employees_df.salary.desc()).limit(3).show()
```

PART 6 — Aggregations

Exercise 21 - Find total salary of all employees.

```python
from ast import alias
from pyspark.sql.functions import sum,avg,min,max
employees_df.select(sum("salary").alias("total_salary")).show()
```

Exercise 22 - Find average salary.

```python
employees_df.select(avg("salary").alias("Average_salary")).show()
```

Exercise 23 - Find highest salary.

```python
employees_df.select(max("salary").alias("Highest_salary")).show()
```

Exercise 24 - Find lowest salary.

```python
employees_df.select(min("salary").alias("Lowest_salary")).show()
```

PART 7 — GroupBy

Exercise 25 - Count employees per department.

```python
employees_df.groupBy("department").count().show()
```

Exercise 26 - Calculate average salary per department.

```python
employees_df.groupBy("department").avg("salary").show()
```

Exercise 27 - Find total salary per department.

```python
employees_df.groupBy("department").sum("salary").show()
```

PART 8 — Joining DataFrames

Use employees.csv and departments.csv .

Exercise 28 - Perform inner join.

```python
employees_df.join(departments_df, on="department", how="inner").show()
```

Exercise 29 - Perform left join.

```python
employees_df.join(departments_df, on="department", how="left").show()
```

Exercise 30 - Show employee name with department location.

```python
employees_df.join(departments_df, on="department", how="left") \
    .select("name", "department", "location").show()
```

PART 9 — Updating Data

Exercise 31 - Add a column bonus = salary * 10%.

```python
employees_df.withColumn("bonus", employees_df.salary * 0.10).show()
```

Exercise 32 - Add a column company with value "BotCampus" .

```python
from pyspark.sql.functions import lit
employees_df.withColumn("company", lit("BotCampus")).show()
```

PART 10 — Handling Null Values

Note employee Meera has null department.

Exercise 33 - Find rows with null department.

```python
from pyspark.sql.functions import col
employees_df.filter(col("department").isNull()).show()
```

Exercise 34 - Replace null department with "Unknown" .

```python
employees_df.na.fill({"department": "Unknown"}).show()
```

Exercise 35 - Drop rows with null values.

```python
employees_df.dropna().show()
```

PART 11 — Built-in Functions

Exercise 36 - Convert employee names to uppercase.

```python
from pyspark.sql.functions import upper,lower,length
employees_df.select("name",upper("name").alias("upper_name")).show()
```

Exercise 37 - Find length of employee names.

```python
employees_df.select("name",length("name").alias("name_length")).show()
```

Exercise 38 - Create column salary_in_lakhs .

```python
employees_df.withColumn("salary_in_lakhs", employees_df.salary / 100000).show()
```
