86. Rank customers by balance within city.

```python
spark.sql("""
SELECT c.name,c.city,a.balance,
      RANK() OVER(PARTITION BY c.city ORDER BY a.balance DESC ) as rank_in_city
      FROM customers c
      join accounts a on c.customer_id=a.customer_id
      """).show()
```

87. Use row_number to find top balance account per city.

```python
spark.sql("""
SELECT name, city, balance
FROM (
    SELECT c.name, c.city, a.balance,
           ROW_NUMBER() OVER(PARTITION BY c.city ORDER BY a.balance DESC) AS row_num
    FROM customers c
    JOIN accounts a ON c.customer_id = a.customer_id
)
WHERE row_num = 1
""").show()
```

88. Use dense_rank to rank transaction amount by transaction type.

```python
spark.sql("""
SELECT txn_id, account_id, txn_type, amount,
       DENSE_RANK() OVER(PARTITION BY txn_type ORDER BY amount DESC) AS txn_rank
FROM transactions
""").show()
```

89. Find top 2 transactions per account.

```python
spark.sql("""
SELECT account_id, txn_id, txn_type, amount
FROM (
    SELECT account_id, txn_id, txn_type, amount,
           ROW_NUMBER() OVER(PARTITION BY account_id ORDER BY amount DESC) AS row_num
    FROM transactions
)
WHERE row_num <= 2
""").show()
```

90. Create running transaction total per account_id.

```python
spark.sql("""
SELECT account_id, txn_id, txn_date, amount,
       SUM(amount) OVER(PARTITION BY account_id ORDER BY txn_date) AS running_total
FROM transactions
""").show()
```

91. Rank branches by total balance.

```python
spark.sql("""
SELECT branch, total_balance,
       RANK() OVER(ORDER BY total_balance DESC) AS branch_rank
FROM (
    SELECT branch, SUM(balance) AS total_balance
    FROM accounts
    GROUP BY branch
)
""").show()
```

92. Rank cities by total customer balance.

```python
spark.sql("""
SELECT city, total_balance,
       RANK() OVER(ORDER BY total_balance DESC) AS city_rank
FROM (
    SELECT c.city, SUM(a.balance) AS total_balance
    FROM customers c
    JOIN accounts a ON c.customer_id = a.customer_id
    GROUP BY c.city
)
""").show()
```

93. Rank customers by total transaction amount.

```python
spark.sql("""
SELECT name, total_transaction_amount,
       RANK() OVER(ORDER BY total_transaction_amount DESC) AS customer_rank
FROM (
    SELECT c.name, SUM(t.amount) AS total_transaction_amount
    FROM customers c
    JOIN accounts a ON c.customer_id = a.customer_id
    JOIN transactions t ON a.account_id = t.account_id
    GROUP BY c.name
)
""").show()
```

94. Find highest transaction customer per city.

```python
spark.sql("""
SELECT name, city, amount
FROM (
    SELECT c.name, c.city, t.amount,
           ROW_NUMBER() OVER(PARTITION BY c.city ORDER BY t.amount DESC) AS row_num
    FROM customers c
    JOIN accounts a ON c.customer_id = a.customer_id
    JOIN transactions t ON a.account_id = t.account_id
)
WHERE row_num = 1
""").show()
```

95. Find highest balance customer per region.

```python
spark.sql("""
SELECT name, region, balance
FROM (
    SELECT c.name, b.region, a.balance,
           ROW_NUMBER() OVER(PARTITION BY b.region ORDER BY a.balance DESC) AS row_num
    FROM customers c
    JOIN accounts a ON c.customer_id = a.customer_id
    JOIN branches b ON a.branch = b.branch
)
WHERE row_num = 1
""").show()
```
