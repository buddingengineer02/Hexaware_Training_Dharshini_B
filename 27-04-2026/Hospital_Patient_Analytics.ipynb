86. Rank patients by age within city.

```python
spark.sql("""
SELECT name,age,city,
RANK() OVER(PARTITION BY city ORDER BY age DESC)
AS rank_age from patients
""").show()
```

87. Use row_number to find oldest patient per city.

```python
spark.sql("""
SELECT name, city, age
FROM (
    SELECT name, city, age,
           ROW_NUMBER() OVER(PARTITION BY city ORDER BY age DESC) AS row_num
    FROM patients
)
WHERE row_num = 1
""").show()
```

88. Use dense_rank to rank doctors by fee within specialization.

```python
spark.sql("""
SELECT doctor_name, specialization, consultation_fee,
       DENSE_RANK() OVER(PARTITION BY specialization ORDER BY consultation_fee DESC) AS fee_rank
FROM doctors
""").show()
```

89. Find highest fee doctor per specialization.

```python
spark.sql("""
SELECT doctor_name, specialization, consultation_fee
FROM (
    SELECT doctor_name, specialization, consultation_fee,
           ROW_NUMBER() OVER(PARTITION BY specialization ORDER BY consultation_fee DESC) AS row_num
    FROM doctors
)
WHERE row_num = 1
""").show()
```

90. Find top 2 oldest patients per city.

```python
spark.sql("""
SELECT name, city, age
FROM (
    SELECT name, city, age,
           ROW_NUMBER() OVER(PARTITION BY city ORDER BY age DESC) AS row_num
    FROM patients
)
WHERE row_num <= 2
""").show()
```

91. Rank doctors by consultation fee overall.

```python
spark.sql("""
SELECT doctor_name, specialization, consultation_fee,
       RANK() OVER(ORDER BY consultation_fee DESC) AS overall_fee_rank
FROM doctors
""").show()
```

92. Rank cities by patient count.

```python
spark.sql("""
SELECT city, patient_count,
       RANK() OVER(ORDER BY patient_count DESC) AS city_rank
FROM (
    SELECT city, COUNT(*) AS patient_count
    FROM patients
    GROUP BY city
)
""").show()
```

93. Rank doctors by appointment count.

```python
spark.sql("""
SELECT doctor_name, appointment_count,
       RANK() OVER(ORDER BY appointment_count DESC) AS appointment_rank
FROM (
    SELECT d.doctor_name, COUNT(a.appointment_id) AS appointment_count
    FROM doctors d
    JOIN appointments a ON d.doctor_id = a.doctor_id
    GROUP BY d.doctor_name
)
""").show()
```

94. Create running total bill amount by payment mode.
