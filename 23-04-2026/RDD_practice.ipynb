RDD-Resilient Distributed Dataset

A distributed datastructure in Spark

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("RDD Basics").getOrCreate()
sc=spark.sparkContext
print("spark Ready")
```

```python
numbers=[10,20,30,40,50]
rdd=sc.parallelize(numbers)
print(rdd)
```

```python
rdd.collect()
```

```python
data=["Rahul","Sneha","Arjun","Priya"]
name_rdd=sc.parallelize(data)
print(name_rdd)
```

```python
name_rdd.collect()
```

Lazy Evaluation - transformations are not executed immediately but are delayed until an action is triggered.

RDD operations: two categories
1. Transformations
These create a new RDD.
Examples:
map,
filter,
flatMap,
distinct,
union,
intersection,
groupByKey,
reduceByKey
2. Actions
These return a result to the driver.
Examples:
collect,
count,
first,
take,
reduce

TRANSFORMATIONS

```python
rdd=sc.parallelize([1,2,3,4,5])
mapped_rdd=rdd.map(lambda x:x*2)
filtered_rdd=mapped_rdd.filter(lambda x: x>5)
print("Nothing Executed Yet")
```

```python
filtered_rdd.collect()
```

map - Allows to Iterate

```python
rdd=sc.parallelize([1,2,3,4,5])
squared_rdd=rdd.map(lambda x:x*x)
squared_rdd.collect()
```

```python
rdd=sc.parallelize(["rahul","sneha","arjun"])
upper_rdd=rdd.map(lambda x:x.upper())
upper_rdd.collect()
```

filter - Keep the matching element

```python
rdd=sc.parallelize([10,15,20,25,30])
even_rdd=rdd.filter(lambda x:x%2==0)
even_rdd.collect()
```

```python
rdd=sc.parallelize(["apple","banana","mango","kiwi"])
long_words=rdd.filter(lambda x:len(x)>5)
long_words.collect()
```

flatmap - similar to map but we can catch  multiple elements

```python
lines=sc.parallelize([
    "python spark hadoop",
    "spark sql pyspark"
    ])
words=lines.flatMap(lambda line:line.split(" "))
words.collect()
```

map vs flatMap

```python
lines=sc.parallelize(["hwllo world","pyspark rdd"])
result=lines.map(lambda x:x.split(" "))
result.collect()
```

```python
lines=sc.parallelize(["hwllo world","pyspark rdd"])
result=lines.flatMap(lambda line:line.split(" "))
result.collect()
```

distinct - Unique elements

```python
rdd=sc.parallelize([10,10,20,20,30,40,50,60,60])
rdd.distinct().collect()
```

Union - combines two rdds

```python
rdd1=sc.parallelize([1,2,3])
rdd2=sc.parallelize([4,5,6])
rdd1.union(rdd2).collect()
```

Intersection -  common elements

```python
rdd1=sc.parallelize([1,2,3,4,5])
rdd2=sc.parallelize([4,5,6,7])
rdd1.intersection(rdd2).collect()
```

Subtract

```python
rdd1=sc.parallelize([1,2,3,4])
rdd2=sc.parallelize([3,4])
rdd1.subtract(rdd2).collect()
```

```python
rdd1=sc.parallelize([1,2,3,4])
rdd2=sc.parallelize([3,4,5,6])
rdd2.subtract(rdd1).collect()
```

reduce - reduce the data into one number

```python
rdd=sc.parallelize([10,20,30,40])
rdd.reduce(lambda a,b:a+b)
```

Key value Pair

reduceByKey - Cummulative output

```python
data=[("IT",70000),("HR",60000),("IT",75000),("Finance",80000),("HR",58000)]
rdd=sc.parallelize(data)
rdd.collect()
```

```python
rdd.reduceByKey(lambda a,b:a+b).collect()
```

groupByKey - Splitted output

```python
grouped=rdd.groupByKey()
[(k,list(v)) for k,v in grouped.collect()]
```

mapValues - only works on values not keys

```python
rdd=sc.parallelize([
    ("Rahul",85),
    ("Sneha",92),
    ("Arjun",78)
])
rdd.mapValues(lambda x:x+5).collect()
```

sortBy

```python
rdd=sc.parallelize([10,40,50,30,20])
rdd.sortBy(lambda x:x).collect()
```

Word count

```python
lines=sc.parallelize([
    "python spark python",
    "spark hadoop spark",
    "python hadoop"
])
word_counts=(
    lines
    .flatMap(lambda line: line.split(" "))
    .map(lambda word: (word,1))
    .reduceByKey(lambda a,b:a+b)
    )
word_counts.collect()
```

```python
employees=sc.parallelize([
    ("Rahul",70000),
    ("Sneha",60000),
    ("Arjun",75000),
    ("Priya",80000),
    ("Karan",50000)
])
high_salary=employees.filter(lambda x:x[1]>65000)
high_salary.collect()
```
