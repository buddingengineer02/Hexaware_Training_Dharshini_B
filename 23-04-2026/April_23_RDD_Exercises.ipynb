Dataset 1 — Website Logs

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("RDD Basics").getOrCreate()
sc=spark.sparkContext
print("spark Ready")
```

```python
logs = [
"Rahul login",
"Sneha login",
"Arjun login",
"Rahul purchase",
"Priya login",
"Sneha logout",
"Rahul login",
"Karan login",
"Arjun purchase",
"Sneha login",
"Rahul logout",
"Amit login",
"Priya purchase",
"Karan purchase",
"Sneha login",
"Rahul purchase",
"Meera login",
"Arjun login",
"Sneha purchase",
"Rahul login"
]
logs_rdd = sc.parallelize(logs)
logs_rdd.collect()
```

Part 1 — Basic RDD Exercises

Using logs_rdd .

Exercise 1 - Count total log entries.

```python
logs_rdd.count()
```

Exercise 2 - Print first 5 log records.

```python
logs_rdd.take(5)
```

Exercise 3 - Extract only user names.

```python
user_names=logs_rdd.map(lambda x:x.split(" ")[0])
user_names.collect()
```

Exercise 4 - Extract only actions.

```python
actions=logs_rdd.map(lambda x:x.split(" ")[1])
actions.collect()
```

Exercise 5 - Find unique users.

```python
user_names=logs_rdd.map(lambda x:x.split(" ")[0]).distinct()
user_names.collect()
```

Exercise 6 - Count number of login actions.

```python
login_count=logs_rdd.filter(lambda x:x.split()[1]=="login").count()
login_count
```

Exercise 7 - Find all purchase actions.

```python
purchase= logs_rdd.filter(lambda x: x.split()[1] == "purchase")
purchase.collect()
```

Part 2 — flatMap Practice

Create word dataset.

```python
sentences = sc.parallelize([

"big data spark",

"spark hadoop hive",

"python spark hadoop",

"spark streaming kafka" ])
```

Exercise 8 - Split all sentences into words.

```python
words = sentences.flatMap(lambda x: x.split())
words.collect()
```

Exercise 9 - Find all distinct technologies.

```python
distinct_tech = sentences.flatMap(lambda x: x.split()).distinct()
distinct_tech.collect()
```

Exercise 10

Count number of occurrences of each word.

Expected result format:

spark → 4

hadoop → 2

python → 1

```python
word_count = sentences.flatMap(lambda x: x.split()) \
                      .map(lambda x: (x, 1)) \
                      .reduceByKey(lambda a, b: a + b)

word_count.collect()
```

Dataset 2 — Product Sales

```python
sales = [
("Laptop",1,75000),
("Mouse",3,500),
("Keyboard",2,1500),
("Laptop",1,75000),
("Monitor",1,12000),
("Mouse",2,500),
("Keyboard",1,1500),
("Laptop",1,75000),
("Monitor",2,12000),
("Mouse",3,500),
("Laptop",1,75000),
("Keyboard",2,1500),
("Monitor",1,12000),
("Mouse",1,500),
("Laptop",1,75000)
]
sales_rdd = sc.parallelize(sales)
sales_rdd.collect()
```

Exercise 11 - Calculate revenue per row.

```python
revenue= sales_rdd.map(lambda x: (x[0], x[1] * x[2]))
revenue.collect()
```

Exercise 12 - Calculate total revenue of all sales.

```python
total_revenue = sales_rdd.map(lambda x: x[1] * x[2]).reduce(lambda a, b: a + b)
total_revenue
```

Exercise 13 - Find distinct products sold.

```python
distinct_products = sales_rdd.map(lambda x: x[0]).distinct()
distinct_products.collect()
```

Exercise 14 - Find all products with quantity greater than 1.

```python
qty_greater_one = sales_rdd.filter(lambda x: x[1] > 1)
qty_greater_one.collect()
```

Exercise 15- Calculate total quantity sold per product.

Expected format:

Laptop → 5

Mouse → 9

Keyboard → 5

```python
total_quantity = sales_rdd.map(lambda x: (x[0], x[1])) \
                          .reduceByKey(lambda a, b: a + b)

total_quantity.collect()
```

Exercise 16

Calculate total revenue per product.

Expected output example:

Laptop → 375000

Mouse → 4500

Keyboard → 7500

Monitor → 48000

```python
total_revenue_product = sales_rdd.map(lambda x: (x[0], x[1] * x[2])) \
                                 .reduceByKey(lambda a, b: a + b)

total_revenue_product.collect()
```

Exercise 17 - Find the highest revenue product.

```python
highest_revenue_product = sales_rdd.map(lambda x: (x[0], x[1] * x[2])) \
                                   .reduceByKey(lambda a, b: a + b) \
                                   .sortBy(lambda x: x[1], ascending=False) \
                                   .take(1)

highest_revenue_product
```

Dataset 3 — Employee Data

```python
employees = [
("Rahul","IT",70000),
("Sneha","HR",60000),
("Arjun","IT",75000),
("Priya","Finance",80000),
("Karan","IT",50000),
("Amit","HR",58000),
("Meera","Finance",82000),
("Ravi","IT",72000),
("Neha","HR",61000),

("Vikram","Finance",90000)
]
emp_rdd = sc.parallelize(employees)
emp_rdd.collect()
```

Part 4 — Employee Dataset

Using emp_rdd .

Exercise 18 - Extract only employee names.

```python
emp_names = emp_rdd.map(lambda x: x[0])
emp_names.collect()
```

Exercise 19 - Find employees with salary greater than 70,000.

```python
sal_gt_70000 = emp_rdd.filter(lambda x: x[2] > 70000)
sal_gt_70000.collect()
```

Exercise 20 - Find distinct departments.

```python
departments = emp_rdd.map(lambda x: x[1]).distinct()
departments.collect()
```

Exercise 21

Calculate total salary per department.

Expected output:

IT → total salary

HR → total salary

Finance → total salary

```python
salary_per_dept = emp_rdd.map(lambda x: (x[1], x[2])) \
                               .reduceByKey(lambda a, b: a + b)

salary_per_dept.collect()
```

Exercise 22 - Calculate average salary per department.

```python
avg_salary_dept = emp_rdd.map(lambda x: (x[1], (x[2], 1))) \
                         .reduceByKey(lambda a, b: (a[0] + b[0], a[1] + b[1])) \
                         .mapValues(lambda x: x[0] / x[1])

avg_salary_dept.collect()
```

Exercise 23 - Find the highest paid employee.

```python
highest_paid = emp_rdd.sortBy(lambda x: x[2], ascending=False).take(1)
highest_paid
```

Part 5 — Intermediate RDD Exercises

Exercise 24 - Count number of employees per department.

Expected output:

IT → 4

HR → 3

Finance → 3

```python
emp_count_dept = emp_rdd.map(lambda x: (x[1], 1)) \
                        .reduceByKey(lambda a, b: a + b)
emp_count_dept.collect()
```

Exercise 25 - Sort employees by salary descending.

```python
sorted_employees = emp_rdd.sortBy(lambda x: x[2], ascending=False)
sorted_employees.collect()
```

Exercise 26 - Find top 3 highest salaries.

```python
top_3_salaries = emp_rdd.map(lambda x: x[2]).sortBy(lambda x: x, ascending=False).take(3)
top_3_salaries
```

Exercise 27 - Create (department, employee_name) pairs.

```python
dept_emp_pairs = emp_rdd.map(lambda x: (x[1], x[0]))
dept_emp_pairs.collect()
```

Exercise 28 - Group employees by department.

```python
grouped_dept = emp_rdd.map(lambda x: (x[1], x[0])).groupByKey() \
                      .mapValues(list)

grouped_dept.collect()
```

Part 6 — Advanced RDD Thinking

Exercise 29 - Find department with highest total salary.

```python
highest_salary_dept = emp_rdd.map(lambda x: (x[1], x[2])) \
                             .reduceByKey(lambda a, b: a + b) \
                             .sortBy(lambda x: x[1], ascending=False) \
                             .take(1)

highest_salary_dept
```

Exercise 30 - Find employee count and total salary per department.

Expected output format:

IT → (count, total_salary)

```python
count_total_salary = emp_rdd.map(lambda x: (x[1], (1, x[2]))) \
                            .reduceByKey(lambda a, b: (a[0] + b[0], a[1] + b[1]))

count_total_salary.collect()
```

Final Mini Project (RDD)

Use sales dataset.

Tasks:

1. Calculate total revenue per product

```python
product_revenue = sales_rdd.map(lambda x: (x[0], x[1] * x[2])) \
                           .reduceByKey(lambda a, b: a + b)

product_revenue.collect()
```

2. Find top selling product

```python
top_selling_product = sales_rdd.map(lambda x: (x[0], x[1])) \
                               .reduceByKey(lambda a, b: a + b) \
                               .sortBy(lambda x: x[1], ascending=False) \
                               .take(1)

top_selling_product
```

3. Find product generating highest revenue

```python
highest_revenue_product = sales_rdd.map(lambda x: (x[0], x[1] * x[2])) \
                                   .reduceByKey(lambda a, b: a + b) \
                                   .sortBy(lambda x: x[1], ascending=False) \
                                   .take(1)

highest_revenue_product
```

4. Find products sold more than 3 times

```python
products_more_than_3 = sales_rdd.map(lambda x: (x[0], x[1])) \
                                .reduceByKey(lambda a, b: a + b) \
                                .filter(lambda x: x[1] > 3)

products_more_than_3.collect()
```

5. Sort products by revenue

```python
sorted_products_by_revenue = sales_rdd.map(lambda x: (x[0], x[1] * x[2])) \
                                      .reduceByKey(lambda a, b: a + b) \
                                      .sortBy(lambda x: x[1], ascending=False)

sorted_products_by_revenue.collect()
```
