Section 5 — Conditional Queries

26. Show employees with salary greater than company average.

```sql
%sql
select * 
from employees 
where salary>(select avg(salary) 
from employees)
```

27. Show employees with experience greater than department average.

```sql
%sql
SELECT *
FROM employees e
WHERE experience > (
    SELECT AVG(experience)
    FROM employees
    WHERE department = e.department
);
```

Section 6 — Joins

31. Show employees with project budgets by joining employees and projects tables using
department.

```sql
%sql
select
 e.emp_name,p.project_name,e.department,p.budget 
from employees e 
JOIN projects p 
on e.department=p.department;
```

32. Show employee name, department, project name.

```sql
%sql
select e.emp_name, e.department, p.project_name
from employees e
JOIN projects p
ON e.department = p.department;
```

33. Show department-wise project budget.

```sql
%sql
select p.department, SUM(p.budget) AS total_budget
from projects p
GROUP BY p.department;
```

34. Show employees working in departments that have projects.

```sql
%sql
select DISTINCT e.*
from employees e
join projects p
ON e.department = p.department;
```

35. Count number of employees working per project.

```sql
%sql
select p.project_name, COUNT(e.emp_id) AS total_employees
from projects p
JOIN employees e
ON p.department = e.department
GROUP BY p.project_name;
```

Section 7 — Advanced Queries

36. Find top 3 highest paid employees in Engineering department.

```sql
%sql
select *
from employees
where department = 'Engineering'
ORDER BY salary DESC
LIMIT 3;
```

37. Find department with highest average salary.


```sql
%sql
select department, AVG(salary) AS avg_salary
from employees
GROUP BY department
ORDER BY avg_salary DESC
LIMIT 1;
```

38. Find city with highest number of employees.

```sql
%sql
select city, COUNT(*) AS total_employees
FROM employees
GROUP BY city
ORDER BY total_employees DESC
LIMIT 1;
```

39. Find employees whose salary is higher than department average.

```sql
%sql
select *
from employees e
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
    WHERE department = e.department
);
```

40. Find employees whose experience is highest in their department.

```sql
%sql
select *
from employees e
WHERE experience = (
    SELECT MAX(experience)
    FROM employees
    WHERE department = e.department
);
```

```sql
%sql
CREATE OR REPLACE TABLE employee_salary_analysis AS
SELECT 
    emp_name,
    city,
    department,
    salary,
    experience,

    -- Salary Category
    CASE 
        WHEN salary >= 95000 THEN 'High'
        WHEN salary >= 80000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_category,

    -- Department Average Salary
    AVG(salary) OVER (PARTITION BY department) AS department_avg_salary

FROM employees;
```
