Part 8 — Merge / Upsert

71. Create a Delta target table for visits.


```python
daily_visits_data = [
(26,1002,201,"2024-03-14","Completed",2),
(27,1005,210,"2024-03-14","Completed",1),
(11,1011,205,"2024-03-06","Completed",3),
(20,1020,206,"2024-03-10","Completed",2),
(28,1018,203,"2024-03-15","Pending",2)
]

daily_visits_columns = [
"visit_id","patient_id","doctor_id",
"visit_date","visit_status","tests_count"
]

daily_visits_df = spark.createDataFrame(
    daily_visits_data,
    daily_visits_columns
)
```

```python
%sql
CREATE OR REPLACE TABLE visits_target
(
visit_id INT,
patient_id INT,
doctor_id INT,
visit_date STRING,
visit_status STRING,
tests_count INT
)
USING DELTA
```

72. Load initial visits into target table.


```python
%sql
DROP TABLE IF EXISTS visits_target
```

```python
visits_df.write.format("delta") \
.mode("overwrite") \
.saveAsTable("visits_target")
```

```python
%sql
SELECT *
FROM visits_target
```

73. Create a temp view for daily updates.


```python

daily_visits_df.createOrReplaceTempView(
    "daily_updates"
)
```

74. Merge daily updates into the target table.


```python
%sql
MERGE INTO visits_target AS target
USING daily_updates AS source
ON target.visit_id = source.visit_id

WHEN MATCHED THEN
UPDATE SET
target.patient_id = source.patient_id,
target.doctor_id = source.doctor_id,
target.visit_date = source.visit_date,
target.visit_status = source.visit_status,
target.tests_count = source.tests_count

WHEN NOT MATCHED THEN
INSERT
(
visit_id,
patient_id,
doctor_id,
visit_date,
visit_status,
tests_count
)

VALUES
(
source.visit_id,
source.patient_id,
source.doctor_id,
source.visit_date,
source.visit_status,
source.tests_count
)
```

85. Compare Parquet table and Delta table behavior.


| Feature            | Parquet Table              | Delta Table               |
| ------------------ | -------------------------- | ------------------------- |
| Storage Format     | Columnar storage format    | Built on top of Parquet   |
| UPDATE Support     | Not supported directly     | Supported                 |
| DELETE Support     | Not supported directly     | Supported                 |
| MERGE Support      | Not supported              | Supported                 |
| Time Travel        | Not available              | Available                 |
| History Tracking   | Not available              | Available                 |
| ACID Transactions  | Not supported              | Supported                 |
| Schema Enforcement | Limited                    | Strong schema enforcement |
| Performance        | Good                       | Better with optimization  |
| Data Reliability   | Less reliable for updates  | More reliable             |
| Best Use Case      | Raw storage / Landing zone | Processing and analytics  |
